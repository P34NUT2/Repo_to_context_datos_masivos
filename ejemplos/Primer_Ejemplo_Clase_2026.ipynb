```python
from pyspark.sql.functions import avg
df_materias.groupBy("Materia").agg(avg("Calificación").alias("Promedio")).show()
```

```python
from pyspark.sql.functions import col
df_alumnos.join(df_materias, df_alumnos.AlumnoID == df_materias.AlumnoID)\
            .groupBy("Nombre").agg(avg("Calificación").alias("Promedio General"))\
            .where(col("Promedio General") > 85).show()
```

```python
from pyspark.sql.functions import col
df_alumnos.join(df_materias, df_alumnos.AlumnoID == df_materias.AlumnoID)\
            .groupBy("Nombre").agg(avg("Calificación").alias("Promedio General"))\
            .where(col("Promedio General") > 85).show()
```

```python
from pyspark.sql.functions import lit
df_alumnos.select(col("Nombre"),lit("columna creada")).show()
```

```python
df_para_explicar = df_alumnos.join(df_materias, df_alumnos.AlumnoID == df_materias.AlumnoID)\
            .groupBy("Nombre").agg(avg("Calificación").alias("Promedio General"))\
            .where(col("Promedio General") > 85)

df_para_explicar.explain(mode="extended")
```
